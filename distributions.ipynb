```python
import pandas as pd
raw = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
raw["TotalCharges"] = pd.to_numeric(raw["TotalCharges"], errors="coerce")

print(raw["Churn"].value_counts(), raw["Churn"].value_counts(normalize=True).round(3))
print(raw["SeniorCitizen"].mean().round(3))
print(raw["Contract"].value_counts(normalize=True).round(3))
print(raw.groupby("Contract")["Churn"].apply(lambda s: (s == "Yes").mean()).round(3))
print(raw.groupby("InternetService")["Churn"].apply(lambda s: (s == "Yes").mean()).round(3))
print(raw[["tenure", "MonthlyCharges", "TotalCharges"]].describe().round(1))
```
